```python
places=['Lahore','Islambad','Najaf','Karbala','Karachi']
print(places)
print("Places after sorted:",sorted(places))                                  
print("original places list:",places)
print("Places after reverse sort:",sorted(places, reverse=True))
print("original places list:",places)
places=['Lahore','Islambad','Najaf','Karbala','Karachi']
places.reverse()
print("Places after first reverse:",places)
places.reverse()
print("Places after second reverse:",places)
places.sort()
print("Places after sorting alphabetically:",places)
places.sort(reverse=True)
print("Places after sorting reverse alphabetically:",places)
```
